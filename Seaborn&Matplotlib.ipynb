Function to add short observations under each plot explaining insights.

```python
def add_observation(text, y=-0.15):
    plt.figtext(0.5, y, text, ha="center", fontsize=10)
```

Countplot: Passenger Count by Embarked Port

Create a countplot to display the number of passengers who boarded from each port (embarked), separated by sex.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the Titanic dataset from Seaborn
titanic = sns.load_dataset("titanic")

# Set plot style
sns.set(style="whitegrid")

# Create the countplot
plt.figure(figsize=(8, 5))
sns.countplot(
    data=titanic,
    x="embarked",
    hue="sex",
    palette="Set2"
)

# Add labels and title
plt.title("Passenger Count by Embarked Port and Sex")
plt.xlabel("Embarked Port")
plt.ylabel("Number of Passengers")

add_observation("Number of Male Passengers are more than Female Passengers")

# Show the plot
plt.show()
```

Swarmplot: Age Distribution across Classes

Create a swarmplot to visualize the distribution of age across passenger classes (pclass), categorized by sex.

```python
# Remove rows with missing age values (required for swarmplot)
titanic_age = titanic.dropna(subset=["age"])

# Set plot style
sns.set(style="whitegrid")

# Create the swarmplot
plt.figure(figsize=(10, 6))
sns.swarmplot(
    data=titanic_age,
    x="pclass",
    y="age",
    hue="sex",
    palette="Set1",
    dodge=True
)

# Add labels and title
plt.title("Age Distribution across Passenger Classes by Sex")
plt.xlabel("Passenger Class")
plt.ylabel("Age")

# Adjust legend
plt.legend(title="Sex")

add_observation("Third-class passengers include many younger individuals and children")

# Show the plot
plt.show()
```

Pie Chart: Gender Distribution

Create a pie chart using Matplotlib to show the percentage of male
vs. female passengers. - Add percentage labels using autopct - Use explode for one slice to highlight it.

```python
# Count male and female passengers
gender_counts = titanic["sex"].value_counts()

# Data for pie chart
labels = gender_counts.index
sizes = gender_counts.values

# Explode one slice (highlight females)
explode = (0.1, 0)  # explode first slice only

# Create the pie chart
plt.figure(figsize=(6, 6))
plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
    explode=explode,
    colors=["lightcoral", "lightskyblue"]
)

# Add title
plt.title("Gender Distribution of Titanic Passengers")

# Equal aspect ratio ensures the pie is circular
plt.axis("equal")

add_observation("Majority of the passengers are Males")

# Show the plot
plt.show()
```

Heatmap: Correlation Matrix

Create a heatmap to show correlations between age, fare, pclass, sibsp, parch, and survived. (Use seaborn.heatmap with annotations)

```python
# Select relevant columns
cols = ["age", "fare", "pclass", "sibsp", "parch", "survived"]

# Drop rows with missing values in selected columns
titanic_corr = titanic[cols].dropna()

# Compute correlation matrix
corr_matrix = titanic_corr.corr()

# Set plot style
sns.set(style="white")

# Create heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

# Add title
plt.title("Correlation Heatmap of Titanic Variables")

add_observation("Fare and passenger class show a strong negative correlation")

# Show plot
plt.show()
```

Violin Plot: Fare by Class and Gender

Create a violin plot to visualize fare distribution across classes, separated by gender.

```python
# Set plot style
sns.set(style="whitegrid")

# Create the violin plot
plt.figure(figsize=(10, 6))
sns.violinplot(
    data=titanic,
    x="pclass",
    y="fare",
    hue="sex",
    palette="Set2",
    split=True
)

# Add labels and title
plt.title("Fare Distribution by Passenger Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")

# Adjust legend
plt.legend(title="Sex")

add_observation("First-class has the highest Fare distributions.")

# Show the plot
plt.show()
```

Scatter Plot: Age vs Fare

Create a scatter plot to explore the relationship between age and fare, colored by survival status.

Use a seaborn scatterplot with hue='survived'. Each plot must have titles, labels, and legends where needed.

```python
# Drop rows with missing age or fare values
titanic_scatter = titanic.dropna(subset=["age", "fare"])

# Set plot style
sns.set(style="whitegrid")

# Create the scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=titanic_scatter,
    x="age",
    y="fare",
    hue="survived",
    palette="Set1",
    alpha=0.7
)

# Add labels and title
plt.title("Age vs Fare by Survival Status")
plt.xlabel("Age")
plt.ylabel("Fare")

# Adjust legend labels
plt.legend(title="Survived", labels=["No", "Yes"])


# Show plot
plt.show()
```
