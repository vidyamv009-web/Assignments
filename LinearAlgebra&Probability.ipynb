Load the Dataset & Handle Missing Values

```python
import pandas as pd
import numpy as np
# Load dataset
df = pd.read_csv("/content/sample_data/winequality-red.csv", delimiter=';')
df.head()
```

```python
#Check missing values
df.isnull().sum()
#Replace null values with column mean
df = df.fillna(df.mean())
```

Extract Columns as Vectors: alcohol and citric acid

```python
alcohol_vector = df['alcohol'].values
citric_acid_vector = df['citric acid'].values

print(alcohol_vector[:5])
print(citric_acid_vector[:5])
```

Covariance Matrix of Selected Features

```python
X = df[['alcohol', 'density']].values

cov_matrix = np.cov(X.T)
cov_matrix
```

Eigen Decomposition of Covariance Matrix

```python
#Compute eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
eigenvalues, eigenvectors
```

```python
#Identify top 2 eigenvalues
sorted_indices = np.argsort(eigenvalues)[::-1]
top_eigenvalues = eigenvalues[sorted_indices]
top_eigenvectors = eigenvectors[:, sorted_indices]

top_eigenvalues, top_eigenvectors
```

Most Common Wine Quality & Distribution Interpretation

```python
quality_counts = df['quality'].value_counts()
quality_counts
```

```python
most_common_quality = quality_counts.idxmax()
most_common_quality
```
